```python
import random
from datetime import datetime, timedelta
from pyspark.sql import types as T

# Reproduzierbare Testdaten
random.seed(42)

# Steuergeräte
ecus = [
    "ECU_Engine",
    "ECU_Battery",
    "ECU_Brake",
    "ECU_Charging",
    "ECU_Thermal",
    "ECU_Gateway"
]

# Synthetische Fehlerkategorien
error_categories = [
    "Short_Circuit",
    "Overtemperature",
    "Communication_Loss",
    "Voltage_Deviation",
    "Sensor_Failure",
    "Diagnostic_Error"
]

# Beispielhafte DTC-Codes
dtc_codes = [
    "P0300",
    "P0420",
    "P0A80",
    "C0035",
    "U0100",
    "U0121",
    "P0562",
    "P0118"
]

start_date = datetime(2026, 1, 1)

records = []

for i in range(1, 1001):

    ecu = random.choice(ecus)
    error_category = random.choice(error_categories)
    dtc = random.choice(dtc_codes)

    # Ungefähr 80 % PASS und 20 % FAIL
    result = random.choices(
        ["PASS", "FAIL"],
        weights=[80, 20],
        k=1
    )[0]

    timestamp = start_date + timedelta(
        minutes=random.randint(0, 300000)
    )

    records.append((
        i,
        ecu,
        dtc,
        error_category,
        result,
        timestamp
    ))

schema = T.StructType([
    T.StructField("TestID", T.IntegerType(), False),
    T.StructField("ECU", T.StringType(), False),
    T.StructField("DTC", T.StringType(), False),
    T.StructField("ErrorCategory", T.StringType(), False),
    T.StructField("TestResult", T.StringType(), False),
    T.StructField("TestTimestamp", T.TimestampType(), False)
])

df_synthetic = spark.createDataFrame(
    records,
    schema=schema
)

print("Anzahl generierter Testfälle:", df_synthetic.count())

display(df_synthetic.limit(10))

```
