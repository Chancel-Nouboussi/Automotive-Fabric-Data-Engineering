```python

from pyspark.sql import SparkSession
from datetime import datetime

# Synthetische Automotive-Diagnosedaten
test_data = [
    (1001, "ECU_Engine", "0xP0300", "PASS", "2026-10-20 08:30:00"),
    (1002, "ECU_Battery", "0xP0A80", "FAIL", "2026-10-20 09:15:00"),
    (1003, "ECU_Brake", "0xC0035", "PASS", "2026-10-20 10:00:00"),
    (1004, "ECU_Engine", "0xP0420", "FAIL", "2026-10-21 08:45:00"),
    (1005, "ECU_Battery", "0xP0A80", "PASS", "2026-10-21 11:20:00")
]

columns = [
    "TestID",
    "ECU",
    "DTC",
    "TestResult",
    "TestTimestamp"
]

df_bronze = spark.createDataFrame(test_data, columns)

display(df_bronze)

```

```python

# Bronze Layer – Daten als Delta-Tabelle speichern

df_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("bronze.vehicle_diagnostics")

print("Bronze-Tabelle erfolgreich gespeichert!")
```
